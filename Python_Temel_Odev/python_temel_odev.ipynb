## 2. İç İçe (Nested) if-else

### 2.1 Yaş ve Öğrenci Durumuna Göre Ücret
Kullanıcıdan yaş ve öğrenci olup olmadığı bilgisini alıp; 65 yaş üstü ise "Ücretsiz", öğrenci ise "İndirimli", diğer durumda "Tam ücret" yazdıran bir program.

```
# 2.1 Yaş ve öğrenci durumuna göre ücret belirleme

yas = int(input("Yaşınızı giriniz: "))
ogrenci = input("Öğrenci misiniz? (evet/hayır): ")

if yas > 65:
    print("Ücretsiz")
else:
    if ogrenci == "evet":
        print("İndirimli")
    else:
        print("Tam ücret")
```

```
# 2.2 Sayının hem pozitif hem de çift olup olmadığını kontrol etme

sayi = int(input("Bir sayı giriniz: "))

if sayi > 0:
    if sayi % 2 == 0:
        print("Sayı pozitif ve çifttir.")
    else:
        print("Sayı pozitif ama çift değildir.")
else:
    print("Sayı pozitif değildir.")
```

## 4. for Döngüsü ile Liste Üzerinde İşlem

### 4.1 Listedeki En Büyük Sayıyı Bulma

### 4.2 10'dan Büyük Sayıların Adedini Bulma

```
# 4.1 Listedeki en büyük sayıyı bulma

sayilar = [5, 15, 8, 22, 3, 19]

en_buyuk = sayilar[0]

for sayi in sayilar:
    if sayi > en_buyuk:
        en_buyuk = sayi

print("En büyük sayı:", en_buyuk)
```

```
# 4.2 10'dan büyük sayıların adedini bulma

sayilar = [5, 15, 8, 22, 3, 19]

sayac = 0

for sayi in sayilar:
    if sayi > 10:
        sayac += 1

print("10'dan büyük sayı adedi:", sayac)
```

## 5. while Döngüsü ile Basit Sayaç

### 5.1 10'dan 1'e Geri Sayma

### 5.2 Doğru Şifre Girilene Kadar Tekrar Sorma

```
# 5.1 10'dan 1'e geri sayma

sayac = 10

while sayac >= 1:
    print(sayac)
    sayac -= 1
```

```
# 5.2 Doğru şifre girilene kadar tekrar sorma

dogru_sifre = "1234"

sifre = input("Şifreyi giriniz: ")

while sifre != dogru_sifre:
    print("Şifre yanlış, tekrar deneyiniz.")
    sifre = input("Şifreyi giriniz: ")

print("Şifre doğru.")
```

## 7. Basit Fonksiyon Tanımlama

### 7.2 İki Sayının Toplamını Döndüren Fonksiyon

## 8. Fonksiyon + Koşul Birlikte

### 8.1 Sayının Pozitif, Negatif veya Sıfır Olduğunu Bulma

### 8.2 Üç Sayı Arasından En Büyüğünü Bulma

```
# 8.1 Sayının durumunu bulma

def sayi_durumu(sayi):
    if sayi > 0:
        return "Pozitif"
    elif sayi < 0:
        return "Negatif"
    else:
        return "Sıfır"

sonuc = sayi_durumu(-5)

print("Sayının durumu:", sonuc)
```

```
# 8.2 Üç sayı arasından en büyüğünü bulma

def en_buyuk_bul(sayi1, sayi2, sayi3):
    if sayi1 >= sayi2 and sayi1 >= sayi3:
        return sayi1
    elif sayi2 >= sayi1 and sayi2 >= sayi3:
        return sayi2
    else:
        return sayi3

sonuc = en_buyuk_bul(15, 28, 9)

print("En büyük sayı:", sonuc)
```

## 9. Liste Oluşturma ve İşleme

### 9.1 Kullanıcıdan 5 Şehir Alıp Listeye Ekleme

### 9.2 Listeden Ürün Silme

```
# 9.1 Kullanıcıdan 5 şehir alıp listeye ekleme

sehirler = []

for i in range(5):
    sehir = input("Şehir adı giriniz: ")
    sehirler.append(sehir)

for i, sehir in enumerate(sehirler, start=1):
    print(i, "-", sehir)
```

```
# 9.2 Listeden istenilen ürünü silme

urunler = ["Kalem", "Defter", "Silgi", "Cetvel"]

print("Ürünler:", urunler)

silinecek_urun = input("Silmek istediğiniz ürünü giriniz: ")

if silinecek_urun in urunler:
    urunler.remove(silinecek_urun)
    print("Güncel ürün listesi:", urunler)
else:
    print("Ürün listede bulunamadı.")
```

## 11. Küçük Bir Uygulama — Not Ortalaması Hesaplama

### 11.1 Dört Dersin Not Ortalamasını ve Harf Notunu Hesaplama

### 11.2 En Yüksek ve En Düşük Notu Bulma

```
# 11.1 Dört dersin ortalamasını ve harf notunu hesaplama

not1 = float(input("1. ders notunu giriniz: "))
not2 = float(input("2. ders notunu giriniz: "))
not3 = float(input("3. ders notunu giriniz: "))
not4 = float(input("4. ders notunu giriniz: "))

ortalama = (not1 + not2 + not3 + not4) / 4

if ortalama >= 90:
    harf_notu = "AA"
elif ortalama >= 85:
    harf_notu = "BA"
elif ortalama >= 80:
    harf_notu = "BB"
elif ortalama >= 75:
    harf_notu = "CB"
elif ortalama >= 70:
    harf_notu = "CC"
elif ortalama >= 65:
    harf_notu = "DC"
elif ortalama >= 60:
    harf_notu = "DD"
elif ortalama >= 50:
    harf_notu = "FD"
else:
    harf_notu = "FF"

print("Not ortalaması:", ortalama)
print("Harf notu:", harf_notu)
```

```
# 11.2 En yüksek ve en düşük notu bulma

notlar = [not1, not2, not3, not4]

en_yuksek = max(notlar)
en_dusuk = min(notlar)

print("En yüksek not:", en_yuksek)
print("En düşük not:", en_dusuk)
```

## 12. Küçük Bir Uygulama — Basit Sepet / Alışveriş

### 12.1 100 TL Üzeri Ürünleri Listeleme

### 12.2 Toplam Tutar 500 TL Üzerindeyse %10 İndirim Mesajı Gösterme

```
# 12.1 100 TL üzeri ürünleri listeleme

urunler = {
    "Kalem": 25,
    "Defter": 120,
    "Çanta": 450,
    "Silgi": 20,
    "Kulaklık": 300
}

print("100 TL üzeri ürünler:")

for urun, fiyat in urunler.items():
    if fiyat > 100:
        print(urun, "-", fiyat, "TL")
```

```
# 12.2 Toplam tutara göre indirim kontrolü

fiyatlar = [120, 450, 20, 300]

toplam_tutar = sum(fiyatlar)

print("Toplam tutar:", toplam_tutar, "TL")

if toplam_tutar > 500:
    print("%10 indirim kazandınız")
else:
    print("İndirim kazanamadınız")
```

```
# 12.2 Toplam tutara %10 indirim uygulama
# İndirim uygulandıktan sonra fiyat gösterme

fiyatlar = [120, 450, 20, 300]

toplam_tutar = sum(fiyatlar)

print("Toplam tutar:", toplam_tutar, "TL")

if toplam_tutar > 500:
    indirimli_tutar = toplam_tutar * 0.90
    print("%10 indirim kazandınız")
    print("İndirimli tutar:", indirimli_tutar, "TL")
else:
    print("İndirim kazanamadınız")
```
